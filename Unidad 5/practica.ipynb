```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
import statsmodels.api as sm
```

```python
df1 = pd.read_csv("grupo1.csv")

print("--- ANÁLISIS GRUPO 1 ---")
print(df1.describe())

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df1['Calificacion'], kde=True, color='skyblue')
plt.title('Histograma - Grupo 1')

X = df1[['HorasEstudio']]
y = df1['Calificacion']
modelo = LinearRegression().fit(X, y)
corr, _ = pearsonr(df1['HorasEstudio'], df1['Calificacion'])

plt.subplot(1, 2, 2)
sns.scatterplot(x='HorasEstudio', y='Calificacion', data=df1)
plt.plot(df1['HorasEstudio'], modelo.predict(X), color='red')
plt.title(f'Regresión Grupo 1 (r={corr:.2f})')
plt.show()
```
